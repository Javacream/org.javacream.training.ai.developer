## Aufgabe 2: Input Guardrail

Implementiere eine einfache Eingabeprüfung für offensichtlich unerlaubte oder manipulative Anweisungen.


```java
static Optional<String> rejectReason(String input) {
    if (input == null || input.isBlank()) return Optional.of("Leere Eingabe");
    if (input.length() > 4000) return Optional.of("Eingabe zu lang");
    String normalized = input.toLowerCase();
    if (normalized.contains("ignore previous instructions"))
        return Optional.of("Verdächtige Prompt-Injection-Sequenz");
    return Optional.empty();
}
```
