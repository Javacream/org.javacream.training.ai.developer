## Aufgabe 2: Prompt parametrisieren

Verwende Platzhalter für Thema und Zielgruppe anstatt Strings zusammenzukonkatenieren.


```java
String answer = chatClient.prompt()
    .user(u -> u.text("Erkläre {topic} für {audience} in maximal fünf Sätzen.")
        .param("topic", "Vector Stores")
        .param("audience", "Java-Entwickler"))
    .call()
    .content();
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.


## Aufgabe 3: Structured Output

Lass das Modell eine Lernkarte als Java record mit title, summary und drei keyPoints zurückgeben.


```java
record LearningCard(String title, String summary, java.util.List<String> keyPoints) {}

LearningCard card = chatClient.prompt()
    .user("Erstelle eine Lernkarte zum Thema Prompt Templates.")
    .call()
    .entity(LearningCard.class);

System.out.println(card);
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.


## Aufgabe 4: REST-Endpunkte

Stelle /api/ask synchron und /api/stream als Streaming-Endpunkt bereit.


```java
// Synchron:
@GetMapping("/ask")
String ask(@RequestParam String q) {
    return chatClient.prompt().user(q).call().content();
}

// Streaming:
@GetMapping(value="/stream", produces=MediaType.TEXT_EVENT_STREAM_VALUE)
Flux<String> stream(@RequestParam String q) {
    return chatClient.prompt().user(q).stream().content();
}
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.
