## Aufgabe 1: Memory konfigurieren

Erzeuge ein MessageWindowChatMemory mit einem begrenzten Nachrichtenfenster.


```java
ChatMemory memory = MessageWindowChatMemory.builder()
    .maxMessages(10)
    .build();
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.


## Aufgabe 2: Memory Advisor einbinden

Registriere einen MessageChatMemoryAdvisor am ChatClient.


```java
ChatClient client = ChatClient.builder(chatModel)
    .defaultAdvisors(MessageChatMemoryAdvisor.builder(memory).build())
    .build();
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.


## Aufgabe 3: Conversation IDs

Implementiere einen REST-Endpunkt, der eine conversationId entgegennimmt und getrennte Unterhaltungen ermöglicht.


```java
String answer = client.prompt()
    .user("Mein bevorzugtes Framework ist Spring.")
    .advisors(a -> a.param(ChatMemory.CONVERSATION_ID, "demo-1"))
    .call()
    .content();
```

## Aufgabe 4: Advisor ergänzen

Füge Logging über einen Advisor hinzu und erläutere, warum sensible Inhalte nicht ungefiltert geloggt werden sollten.


```java
ChatClient client = ChatClient.builder(chatModel)
    .defaultAdvisors(
        MessageChatMemoryAdvisor.builder(memory).build(),
        new SimpleLoggerAdvisor())
    .build();
// In Produktion Prompt-/Response-Logging nur nach Datenschutzprüfung aktivieren.
```
