## Aufgabe 1: Read-only Tool erstellen

Implementiere ein Tool, das Produktinformationen aus einem lokalen Katalog liest.


```java
class CatalogTools {
    @Tool(description = "Liest den Preis eines Produkts aus dem lokalen, read-only Seminarkatalog")
    double price(String product) {
        return Map.of("monitor", 299.0, "keyboard", 89.0)
            .getOrDefault(product.toLowerCase(), -1.0);
    }
}
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.


## Aufgabe 2: Parameter beschreiben

Ergänze aussagekräftige Tool- und Parameterbeschreibungen.


```java
@Tool(description = "Liest den Preis eines Produkts aus dem lokalen, read-only Seminarkatalog")
double price(@ToolParam(description = "Produktname, z. B. monitor oder keyboard") String product) {
    return Map.of("monitor", 299.0, "keyboard", 89.0)
        .getOrDefault(product.toLowerCase(), -1.0);
}
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.


## Aufgabe 3: Tool verwenden

Stelle dem Modell eine Frage, die nur mit Hilfe des Tools zuverlässig beantwortet werden kann.


```java
String answer = chatClient.prompt()
    .user("Was kosten zwei Monitore im Seminarkatalog?")
    .tools(new CatalogTools())
    .call()
    .content();
System.out.println(answer);
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.
