## Aufgabe 3: Semantisch suchen

Führe eine Similarity Search für eine Frage aus, deren Wortlaut nicht exakt in den Dokumenten vorkommt.


```java
List<Document> hits = store.similaritySearch(
    SearchRequest.builder()
        .query("Wie nutze ich eigenes Wissen?")
        .topK(2)
        .build());
hits.forEach(System.out::println);
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.


## Aufgabe 4: Retrieval variieren

Vergleiche Ergebnisse für unterschiedliche topK- und similarityThreshold-Werte.


```java
for (double threshold : List.of(0.0, 0.5, 0.8)) {
    var result = store.similaritySearch(
        SearchRequest.builder()
            .query("Wie nutze ich eigenes Wissen?")
            .topK(3)
            .similarityThreshold(threshold)
            .build());
    System.out.println(threshold + " -> " + result.size());
}
```


**Erläuterung:** Die Musterlösung zeigt eine kompakte Referenzimplementierung. Prüfe bei modellabhängigen Aufgaben zusätzlich Fehlerszenarien und alternative Modellantworten.
